```python
import os
os.chdir(os.path.dirname(os.getcwd()))

import json
import hashlib
import pandas as pd
from datetime import datetime

def md5_sha_from_str(val):
    """Generate md5 hash from string"""
    return hashlib.md5(val.encode("utf-8")).hexdigest()
```

```python
df = pd.read_csv('data/evaluation_questions_v2.csv')
```

```python
df.columns = df.columns.str.lower()
```

```python
df['id'] = df['question'].apply(md5_sha_from_str)
```

```python
df[['id','category','script','question','translation']].to_csv('data/evaluation_questions_v2.csv',index=False)
```
